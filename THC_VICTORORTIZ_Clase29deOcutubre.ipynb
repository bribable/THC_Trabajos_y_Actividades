```python

# Clasificación de personas según su edad

edad = int(input("Escribe tu edad: "))

if edad < 0:
    print("La edad ingresada no es válida")
elif edad < 12:
    print("Perteneces a la categoría de niño")
elif edad < 18:
    print("Eres adolescente")
elif edad < 30:
    print("Eres un adulto joven")
elif edad < 65:
    print("Eres adulto")
else:
    print("Eres un adulto mayor")

```

```python

# Programa de ejercicios con condiciones y operadores lógicos

edad = int(input("Escribe tu edad: "))

if edad < 18:
    print("Eres menor de edad")
else:
    print("Eres mayor de edad")


valor = int(input("Introduce un número entero: "))

if valor == 0:
    print("El número es cero")
elif valor > 0:
    print("El número ingresado es positivo")
else:
    print("El número ingresado es negativo")


num = int(input("Escribe un número: "))

if num % 2 != 0:
    print("El número es impar")
else:
    print("El número es par")


estatura = int(input("¿Cuánto mides en centímetros?: "))
anios = int(input("¿Cuántos años tienes?: "))

if estatura < 140 or anios < 12:
    print("No cumples los requisitos para subir")
else:
    print("Puedes subir a la montaña rusa, diviértete")


tipo = input("Ingresa tu tipo de tarifa: ").strip().lower()
edad = int(input("Ingresa tu edad: "))

if tipo == "estudiante" or edad >= 60:
    print("Se aplica la tarifa reducida")
else:
    print("Se aplica la tarifa normal")
```
