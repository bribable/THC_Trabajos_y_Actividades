```python

print("Ingrese su ecuacion cuadratica de la forma ax^2+bx+c")
a = float(input("Ingrese a: "))
b = float(input("Ingrese b: "))
c = float(input("Ingrese c: "))

discriminante = b*b - 4*a*c
Escomplejo = (discriminante < 0)
raizD = discriminante ** (1/2)
x1 = (-b + raizD) / (2 * a)
x2 = (-b - raizD) / (2 * a)

print(f"tu raiz es compleja {Escomplejo}")
print(f"Las soluciones son x1 = {x1} y x2 = {x2}")
```
